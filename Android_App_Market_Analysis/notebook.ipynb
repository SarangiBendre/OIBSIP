# Android App Market Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/apps.csv")
df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
```

Category distribution

```python
df['Category'].value_counts().head(10)
```

Ratings vs installs

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='Rating', y='Installs', data=df)
plt.title("Ratings vs Installs")
plt.show()
```

Free vs Paid apps

```python
df['Type'].value_counts()
```

## Key Insights
- Most apps on the Play Store are free.
- Certain categories dominate the market.
- Higher ratings generally correlate with higher installs.
- Paid apps form a small portion of the market.
